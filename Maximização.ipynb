```python
# ============================================================
# DEMANDAS MARSHALLIANAS - MAXIMIZAÇÃO DE UTILIDADE
# Função Cobb-Douglas
# ============================================================

import sympy as sp

# ------------------------------------------------------------
# 1. DEFININDO AS VARIÁVEIS
# ------------------------------------------------------------

x, y, lamb = sp.symbols('x y lambda', positive=True)

# Parâmetros
A, alpha, beta = sp.symbols('A alpha beta', positive=True)
Px, Py, R = sp.symbols('Px Py R', positive=True)


# ------------------------------------------------------------
# 2. FUNÇÃO DE UTILIDADE
# ------------------------------------------------------------

u = A * x**alpha * y**beta

print("FUNÇÃO DE UTILIDADE:")
display(sp.Eq(sp.Symbol('u'), u))


# ------------------------------------------------------------
# 3. RESTRIÇÃO ORÇAMENTÁRIA
# ------------------------------------------------------------

restricao = Px*x + Py*y - R

print("\nRESTRIÇÃO ORÇAMENTÁRIA:")
display(sp.Eq(Px*x + Py*y, R))


# ------------------------------------------------------------
# 4. LAGRANGIANA
# ------------------------------------------------------------

L = u - lamb * (Px*x + Py*y - R)

print("\nLAGRANGIANA:")
display(sp.Eq(sp.Symbol('L'), L))


# ------------------------------------------------------------
# 5. CONDIÇÕES DE PRIMEIRA ORDEM
# ------------------------------------------------------------

dL_dx = sp.diff(L, x)
dL_dy = sp.diff(L, y)
dL_dlamb = sp.diff(L, lamb)

print("\nCONDIÇÕES DE PRIMEIRA ORDEM:")

print("\n∂L/∂x = 0")
display(sp.Eq(dL_dx, 0))

print("\n∂L/∂y = 0")
display(sp.Eq(dL_dy, 0))

print("\n∂L/∂λ = 0")
display(sp.Eq(dL_dlamb, 0))


# ------------------------------------------------------------
# 6. RESOLVENDO O SISTEMA
# ------------------------------------------------------------

solucao = sp.solve(
    [dL_dx, dL_dy, dL_dlamb],
    [x, y, lamb],
    dict=True
)

print("\nSOLUÇÃO DO SISTEMA:")

for sol in solucao:
    display(sol)


# ------------------------------------------------------------
# 7. DEMANDAS MARSHALLIANAS
# ------------------------------------------------------------

# Demanda Marshalliana do bem X
x_marshall = sp.simplify(
    alpha * R / ((alpha + beta) * Px)
)

# Demanda Marshalliana do bem Y
y_marshall = sp.simplify(
    beta * R / ((alpha + beta) * Py)
)

print("\n================================================")
print("DEMANDAS MARSHALLIANAS")
print("================================================")

print("\nDemanda Marshalliana do bem X:")
display(
    sp.Eq(
        sp.Symbol('x*'),
        x_marshall
    )
)

print("\nDemanda Marshalliana do bem Y:")
display(
    sp.Eq(
        sp.Symbol('y*'),
        y_marshall
    )
)


# ------------------------------------------------------------
# 8. INTERPRETAÇÃO
# ------------------------------------------------------------

print("""
As demandas Marshallianas representam as quantidades ótimas
de X e Y que o consumidor escolhe, considerando:

- preços dos bens (Px e Py);
- renda disponível (R);
- preferências do consumidor (alpha e beta).

Fórmulas:

x* = [alpha / (alpha + beta)] * [R / Px]

y* = [beta / (alpha + beta)] * [R / Py]
""")


# ------------------------------------------------------------
# 9. UTILIDADE MÁXIMA
# ------------------------------------------------------------

u_max = sp.simplify(
    u.subs({
        x: x_marshall,
        y: y_marshall
    })
)

print("\nUTILIDADE MÁXIMA:")
display(
    sp.Eq(
        sp.Symbol('u_max'),
        u_max
    )
)


# ------------------------------------------------------------
# 10. EXEMPLO NUMÉRICO
# ------------------------------------------------------------

valores = {
    A: 1,
    alpha: 0.5,
    beta: 0.5,
    Px: 10,
    Py: 5,
    R: 100
}

x_num = x_marshall.subs(valores)
y_num = y_marshall.subs(valores)
u_num = u_max.subs(valores)

print("\n================================================")
print("EXEMPLO NUMÉRICO")
print("================================================")

print(f"Renda (R): R$ {float(valores[R]):.2f}")
print(f"Preço de X (Px): R$ {float(valores[Px]):.2f}")
print(f"Preço de Y (Py): R$ {float(valores[Py]):.2f}")
print(f"Alpha: {valores[alpha]}")
print(f"Beta: {valores[beta]}")

print("\nDEMANDAS MARSHALLIANAS:")

print(f"x* = {float(x_num):.2f}")
print(f"y* = {float(y_num):.2f}")

print(f"\nUtilidade máxima = {float(u_num):.2f}")


# ------------------------------------------------------------
# 11. VERIFICANDO A RESTRIÇÃO ORÇAMENTÁRIA
# ------------------------------------------------------------

gasto_x = float(valores[Px]) * float(x_num)
gasto_y = float(valores[Py]) * float(y_num)
gasto_total = gasto_x + gasto_y

print("\n================================================")
print("VERIFICAÇÃO DO ORÇAMENTO")
print("================================================")

print(f"Gasto com X: R$ {gasto_x:.2f}")
print(f"Gasto com Y: R$ {gasto_y:.2f}")
print(f"Gasto total: R$ {gasto_total:.2f}")
print(f"Renda disponível: R$ {float(valores[R]):.2f}")

if abs(gasto_total - float(valores[R])) < 0.0001:
    print("\n✓ A restrição orçamentária foi satisfeita!")
else:
    print("\n✗ A restrição orçamentária não foi satisfeita.")
```
